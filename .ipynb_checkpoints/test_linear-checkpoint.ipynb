```python
y_train = train_df['SalePrice']
X_train_raw = train_df.drop(columns=['Id', 'SalePrice'])
X_test_raw  = test_df.drop(columns=['Id'])
```

```python
num_cols = X_train_raw.select_dtypes(include=['int64', 'float64']).columns
cat_cols = X_train_raw.select_dtypes(include=['object']).columns
```

```python
for col in num_cols:
    X_train_raw[col] = X_train_raw[col].fillna(X_train_raw[col].median())
    if col in X_test_raw.columns:
        X_test_raw[col] = X_test_raw[col].fillna(X_train_raw[col].median())
```

```python
X_train_encoded = pd.get_dummies(X_train_raw, drop_first=True)
X_test_encoded  = pd.get_dummies(X_test_raw, drop_first=True)
```

```python

X_train, X_test = X_train_encoded.align(X_test_encoded, join='left', axis=1, fill_value=0)
```
